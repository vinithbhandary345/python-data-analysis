```python
import pandas as pd

# Sample data
data = {
    'Date': ['2021-01-01', '2021-01-01', '2021-01-02', '2021-01-02', '2021-01-03', '2021-01-03'],
    'Region': ['North', 'South', 'North', 'South', 'North', 'South'],
    'Product': ['A', 'A', 'B', 'B', 'C', 'C'],
    'Sales': [100, 150, 200, 130, 300, 250],
    'Quantity': [10, 15, 20, 13, 30, 25]
}

# Create DataFrame
df = pd.DataFrame(data)
```

```python
df
```

```python
# Pivot table with multiple aggregates
pivot_table = pd.pivot_table(
    df,
    values=['Sales', 'Quantity'],
    index=['Date', 'Region'],
    columns=['Product'],
    aggfunc={
        'Sales': ['sum', 'mean'],
        'Quantity': ['sum', 'mean']
    }
)

pivot_table
```
